#Framees Aquarium

```python
import cv2
import pandas as pd
from ultralytics import YOLO
import cvzone
import time
import os
import random

model = YOLO(r'C:\Users\user\Desktop\term 10\res applai\runs\detect\train\weights\best.pt')

cap = cv2.VideoCapture(0)

try:
    with open(r"C:\Users\user\Downloads\vision applai\project\aquarium.v2-release.darknet\test\_darknet.labels", "r") as f:
        class_list = f.read().strip().split("\n")
except FileNotFoundError:
    print("Error: 'label.txt' not found.")
    class_list = []

#print(class_list)
os.makedirs("resf1", exist_ok=True)

frame_count = 0
while cap.isOpened():
    ret, frame = cap.read()
    if not ret:
        print("Failed to capture frame")
        break

    predict_start_time = time.time()
    results = model.predict(frame, verbose=False)
    predict_end_time = time.time()

    predict_duration_ms = (predict_end_time - predict_start_time) * 1000
    print(f"⏱️ {predict_duration_ms:.2f} ms")

    if len(results) > 0 and hasattr(results[0], 'boxes'):
        boxes = results[0].boxes.data
        px = pd.DataFrame(boxes).astype("float")

        for _, row in px.iterrows():
            x1, y1, x2, y2 = map(int, row[:4])
            conf = row[4]  
            class_id = int(row[5]) if len(row) > 5 else -1

            if 0 <= class_id < len(class_list):
                class_name = class_list[class_id]
                print(f"Detected: {class_name} (ID={class_id})")

                color = (random.randint(0, 255), random.randint(0, 255), random.randint(0, 255))
                cv2.rectangle(frame, (x1, y1), (x2, y2), color, 2)
                cvzone.putTextRect(frame, f"{class_name} {conf:.2f}", (x1, y1), scale=1, thickness=1, colorR=color)

    cv2.imshow("Camera", frame)

    frame_count += 1
    output_path = f"resf1/frame_{frame_count:04d}.jpg"
    cv2.imwrite(output_path, frame)
    if cv2.waitKey(1) == ord('q'):
        break

cap.release()
cv2.destroyAllWindows()
print("Done ✅")
```
